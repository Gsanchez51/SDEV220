```python
#Sort 0s, 1s and 2s

class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1
        
        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
                
            elif arr[mid] == 1:
                mid += 1
                
            else:
                arr [mid], arr[high] = arr[high], arr[mid]
                high -= 1

        return arr

arr = [0, 1, 2, 0, 1, 2]

solution = Solution()
print(solution.sort012(arr))
```

```python
#First Occurrence in Sorted

class Solution:
    def firstSearch(self, arr, k):
        low = 0
        high = len(arr) - 1
        result = -1
        
        while low <= high:
            mid = (low + high) // 2
            
            if arr[mid] == k:
                result = mid
                high = mid - 1
                
            elif arr[mid] < k:
                low = mid + 1
                
            else:
                high = mid - 1
                
        return result

arr = [1, 2, 3, 4, 5]
k = 4

solution = Solution()
print(solution.firstSearch(arr, k))
```
